```python
from scipy.stats import f_oneway

# Separate yield data by season
kharif_yield = df_clean[
    df_clean['Season'] == 'Kharif'
]['Yield_Tonnes_Ha'].dropna()

rabi_yield = df_clean[
    df_clean['Season'] == 'Rabi'
]['Yield_Tonnes_Ha'].dropna()

zaid_yield = df_clean[
    df_clean['Season'] == 'Zaid'
]['Yield_Tonnes_Ha'].dropna()

# Perform One-Way ANOVA
f_stat, p_value = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("Yield ANOVA Results")
print("-------------------")
print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Significant difference in yield")
else:
    print("Result: No statistically significant difference in yield")
```

```python
# Profit by season

kharif_profit = df_clean[
    df_clean['Season'] == 'Kharif'
]['Profit_INR'].dropna()

rabi_profit = df_clean[
    df_clean['Season'] == 'Rabi'
]['Profit_INR'].dropna()

zaid_profit = df_clean[
    df_clean['Season'] == 'Zaid'
]['Profit_INR'].dropna()

# One-Way ANOVA
f_stat_profit, p_value_profit = f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("Profit ANOVA Results")
print("--------------------")
print("F-statistic:", f_stat_profit)
print("P-value:", p_value_profit)

if p_value_profit < 0.05:
    print("Result: Significant difference in profit")
else:
    print("Result: No statistically significant difference in profit")
```

```python
# Import required libraries

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Identify numerical and categorical columns

categorical_columns = X.select_dtypes(
    include=['object']
).columns.tolist()

numerical_columns = X.select_dtypes(
    exclude=['object']
).columns.tolist()

print("Categorical columns:", categorical_columns)
print("Numerical columns:", numerical_columns)
```

```python
# Create preprocessing and machine learning pipeline

preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_columns
        ),
        (
            'numerical',
            'passthrough',
            numerical_columns
        )
    ]
)

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', model)
    ]
)

print("Model pipeline created successfully!")
```

```python
# Remove rows where Yield_Tonnes_Ha is missing

df_clean = df.dropna(subset=['Yield_Tonnes_Ha']).copy()

print("Original rows:", len(df))
print("Rows after cleaning:", len(df_clean))
print("Missing yield values:", df_clean['Yield_Tonnes_Ha'].isna().sum())
```

```python
# Define input features and target

X = df_clean.drop(columns=[
    'Farm_ID',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Market_Price_INR_Tonne',
    'Revenue_INR',
    'Total_Cost_INR',
    'Profit_INR',
    'Water_Efficiency_t_per_1000m3'
])

y = df_clean['Yield_Tonnes_Ha']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Split the cleaned dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
# Train the machine learning model

pipeline.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
# Make predictions on test data

y_pred = pipeline.predict(X_test)

print("Predictions completed successfully!")
print("First 5 predictions:", y_pred[:5])
```

```python
# Evaluate the model

mae = mean_absolute_error(y_test, y_pred)

rmse = mean_squared_error(
    y_test,
    y_pred
) ** 0.5

r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")
print("------------------------")
print("MAE:", round(mae, 3))
print("RMSE:", round(rmse, 3))
print("R² Score:", round(r2, 3))
```
